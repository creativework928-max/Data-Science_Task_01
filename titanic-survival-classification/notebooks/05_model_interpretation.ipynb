```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from src.data_loader import load_train_data

df = load_train_data()


# ============================================================
# GENDER
# ============================================================

gender_analysis = (
    df.groupby("Sex")["Survived"]
    .agg(
        Survival_Rate="mean",
        Passenger_Count="count"
    )
    .reset_index()
)

gender_analysis["Survival_Rate"] *= 100

print("\nGENDER ANALYSIS")
print(gender_analysis)


# ============================================================
# PASSENGER CLASS
# ============================================================

class_analysis = (
    df.groupby("Pclass")["Survived"]
    .agg(
        Survival_Rate="mean",
        Passenger_Count="count"
    )
    .reset_index()
)

class_analysis["Survival_Rate"] *= 100

print("\nCLASS ANALYSIS")
print(class_analysis)


# ============================================================
# AGE GROUP
# ============================================================

df["AgeGroup"] = pd.cut(
    df["Age"],
    bins=[
        0,
        12,
        18,
        30,
        45,
        60,
        100
    ],
    labels=[
        "Child",
        "Teen",
        "Young Adult",
        "Adult",
        "Middle Age",
        "Senior"
    ]
)

age_analysis = (
    df.groupby(
        "AgeGroup",
        observed=True
    )["Survived"]
    .agg(
        Survival_Rate="mean",
        Passenger_Count="count"
    )
    .reset_index()
)

age_analysis["Survival_Rate"] *= 100

print("\nAGE ANALYSIS")
print(age_analysis)


# ============================================================
# GENDER + CLASS
# ============================================================

gender_class = (
    df.groupby(
        ["Pclass", "Sex"]
    )["Survived"]
    .mean()
    .reset_index()
)

gender_class["Survival_Rate"] *= 100

print("\nGENDER + CLASS")
print(gender_class)


# ============================================================
# FAMILY SIZE
# ============================================================

df["FamilySize"] = (
    df["SibSp"] +
    df["Parch"] +
    1
)

family_analysis = (
    df.groupby("FamilySize")["Survived"]
    .mean()
    .reset_index()
)

family_analysis["Survival_Rate"] *= 100

print("\nFAMILY SIZE ANALYSIS")
print(family_analysis)


# ============================================================
# SEX + AGE GROUP
# ============================================================

sex_age = (
    df.groupby(
        ["AgeGroup", "Sex"],
        observed=True
    )["Survived"]
    .mean()
    .reset_index()
)

sex_age["Survival_Rate"] *= 100

print("\nSEX + AGE GROUP")
print(sex_age)


# ============================================================
# SEX + CLASS + AGE
# ============================================================

combined = (
    df.groupby(
        ["Pclass", "Sex", "AgeGroup"],
        observed=True
    )["Survived"]
    .agg(
        Survival_Rate="mean",
        Count="count"
    )
    .reset_index()
)

combined["Survival_Rate"] *= 100

print("\nMULTI-FACTOR ANALYSIS")

print(
    combined.sort_values(
        "Survival_Rate",
        ascending=False
    ).head(20)
)
```
